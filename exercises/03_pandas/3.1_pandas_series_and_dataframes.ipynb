## Exercise 1 - Create and access a Pandas Series
Create a pandas series with:
 - the following values:  [1, 3, 2, 7, 5, 4, 9] 
 - indexed with the following fruit names: ["apple", "orange", "cherry", "pear", "pineapple", "mango", "banana"]

Then print:
 - the first three elements of the series
 - all the values associated to a "pear" or a "mango"
 - all the indexes associated to a value greater than 5

````python
import numpy as np
import pandas as pd
````

````python
s1 = pd.Series({'apple': 1, 'orange': 3, 'cherry': 2, 'pear': 7, 'pineapple': 6, 'mango': 4, 'banana': 5})
print(f"Original Series:\n{s1}\n")

print('First three elements:')
print(f"{s1[:3]}\n")  # First three elements

print('Values associated with the keys "pear" or "mango":')
print(f"{s1[['pear', 'mango']]}\n")

print("Index associated to a value greater than 5:")
print(f"{s1[s1 > 5].index}\n")  # Index associated to a value greater than 5
````

## Exercise 2 - Create and access a Pandas Dataframe
Given the following input table with 13 samples and 4 attributes:

```
[[5.1, 3.5, 1, 0.2],
[4.3, 3. , 1, 0.1],
[5. , 0. , 1, 0.4],
[5.1, 3.4, 2, 0.2],
[7.0, 3.2, 1, 0.2],
[6.9, 3.1, 3, 1.5],
[6.7, 3.1, 1, 2. ],
[6. , 2.9, 2, 1.5],
[6.1, 3. , 2, 1.4],
[6.5, 3. , 3, 2.2],
[7.7, 3.8, 3, 2.2],
[7.4, 2.8, 1, 1.9],
[6.8, 3.2, 1, 2.3]]
```

Given the following column names
['height','width','intensity','weight']

Compute the following:
- Create a pandas dataframe with the given data and column names
- Add a new composite feature, 'area' = 'width' * 'height'
- Retrieve the elements associated to an odd index
- Retrieve the 'height' and 'weight' of the samples with area > 20

````python
data = [[5.1, 3.5, 1, 0.2],
        [4.3, 3. , 1, 0.1],
        [5. , 0. , 1, 0.4],
        [5.1, 3.4, 2, 0.2],
        [7.0, 3.2, 1, 0.2],
        [6.9, 3.1, 3, 1.5],
        [6.7, 3.1, 1, 2. ],
        [6. , 2.9, 2, 1.5],
        [6.1, 3. , 2, 1.4],
        [6.5, 3. , 3, 2.2],
        [7.7, 3.8, 3, 2.2],
        [7.4, 2.8, 1, 1.9],
        [6.8, 3.2, 1, 2.3]]

df = pd.DataFrame(data, columns=['height', 'width', 'intensity', 'weight'])
print(df)

# Point 2: add feature area to the DataFrame
df["area"] = df["height"] * df["width"]
print("\nDataFrame with area column:")
print(df)

# Point 3: retrieve the elements associated to an odd index
print("\nElements associated to an odd index:")
df[df.index % 2 == 1]

# Point 4: Retrieve the 'height' and 'weight' of the samples with area > 20
print("\n Height, Weight, Area\n")
print(df[["height", "weight", "area"]])

mask = (df["area"] > 20)
df.loc[mask, ['height', 'weight']]
````
