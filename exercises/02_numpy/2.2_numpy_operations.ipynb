## Exercise 1 Universal Functions
Given an ndarray x = [4, -1, 7, 9, 3, -5] compute the sigmoid function $ y = sigmoid(x)$ which maps numbers $(-\inf +\inf) \rightarrow (0,1)$

$$ sigmoid(x) = \frac{1}{1 + e^{-x}}$$

```python
%pip install numpy
import numpy as np
```

```python

x = np.array([4,-1,7,9,3,-5])
y = 1 / (1 + np.exp(-x))

print(y)
```

## Exercise 2 Aggregation functions
Given an array
x = [
    [2, 3, -4,],
    [7, 5, 9]
]

Return the mean of the elements along the rows and the minimum element along the columns

```python
x = np.array([[2,3,-4], [7,5,9]])

print(x)
print(f"x.shape = {x.shape}")
print(f"dtype = {x.dtype}")

mean = x.mean(axis=-1)
print(f"mean = {mean}")

min = x.min(axis=0)
print(f"min = {min}")
```

## Exercise 3 Sorting
Given the same array as in Ex. 1, i.e., x = [4, -1, 7, 9, 3, -5], find the 3 highest values and the positions of the 2 lowest values

```python
x = np.array([4, -1, 7, 9, 3, -5])

ordine = np.argsort(x)  # indici degli elementi ordinati dal più piccolo al più grande

print(f"ordine = {ordine}")

tre_piu_alti = x[ordine[-3:][::-1]]
posizioni_due_piu_bassi = ordine[:2]

print("Tre valori più alti:", tre_piu_alti)
print("Posizioni dei due valori più bassi:", posizioni_due_piu_bassi)
```

# Exercise 4 Broadcasting
Compute the product between the array x1 = [[0,1,2], [3,4,5]] and the array x2 = [1, 2]
Does it raise an error? How shall x2 be modified to make the operation possible?


```python
x1 = np.array([[0,1,2], [3,4,5]])
print(f"x1 = {x1}")

x2 = np.array([1,2])
print(f"x2 = {x2}")

try:
    prod = x1 * x2
    print(f"prod = {prod}")
except ValueError as e:
    print(f"Errore: {e}")

x2_colonna = x2[:, np.newaxis]  # shape (2, 1)
print(f"x2_colonna = {x2_colonna}")
print(f"shape = {x2_colonna.shape}")

prod = x1 * x2_colonna
print(f"prod = {prod}")
```

## Exercise 5 Broadcasting in practice: centering the features
A dataset is often stored as a matrix with **one row per sample** and **one column per feature**. Given

data = [
    [1, 200],
    [3, 400],
    [5, 900]
]

subtract from every column its mean, so that each feature has mean 0, **without a loop**. Which shapes are involved, and which broadcasting rules apply? Check that the mean of each column of the result is 0.

```python
data = np.array([[1, 200], [3, 400], [5, 900]])
print(f"data =\n{data}")
print(f"shape = {data.shape}")

means = data.mean(axis=0)
print(f"means = {means}")
print(f"shape = {means.shape}")

subtracted = data - means
print(f"subtracted =\n{subtracted}")
print(f"shape = {subtracted.shape}")

means_after = subtracted.mean(axis=0)
print(f"Media delle colonne dopo la sottrazione = {means_after}")
```
